### French Parties

```python
# This dataset was produced by the Parla Gov project. 
# https://www.parlgov.org/
# https://dataverse.harvard.edu/dataverse/parlgov
# for the 2024 election, we are using the view_party.tab file
# for the 2019 election, we are using the The file is view_party.tab.tab file

df_elections = pd.read_csv(f'{DATA_DIR}/parties_description/view_election_2024_editions_all.csv')
```

```python
df_elections_france = df_elections[df_elections['country_name'] == 'France']
# select only the European Parliament elections
df_elections_france_ep = df_elections_france[df_elections_france['election_type'] == 'ep']
# select only the 2019 European Parliament elections
df_elections_france_ep_2019 = df_elections_france_ep[df_elections_france_ep['election_date'] == '2019-05-26']

# select only the columns we need
selected_columns = ['party_id', 'party_name_short', 'party_name',  'party_name_english', 'left_right', 'vote_share', 'seats']
df_elections_france_ep_2019 = df_elections_france_ep_2019[selected_columns]

# remove parties with no seats
df_elections_france_ep_2019_seats = df_elections_france_ep_2019[df_elections_france_ep_2019['seats'] > 0]

df_elections_france_ep_2019_seats.to_csv(f'{DATA_DIR}/parties_description/parlgov_parties_france_2019.csv')
df_elections_france_ep_2019_seats.head(10)
```

```python
df_parties_2024 = pd.read_csv(f'{DATA_DIR}/parties_description/view_party_2024.csv')
df_parties_2024_france = df_parties_2024[df_parties_2024['country_name'] == 'France']
selected_columns = ['party_name_short', 'party_name_english', 'party_name', 'family_name_short', 'family_name', 'left_right']
df_parties_2024_france = df_parties_2024_france[selected_columns]
df_parties_2024_france.rename(columns={'left_right': 'lr_score'}, inplace=True)
df_parties_2024_france.to_csv(f'{DATA_DIR}/parties_description/parlgov_parties_france_2024.csv')
df_parties_2024_france.head(5)
```

## Plot election results

```python
elections = {
    'europe_2019': political_parties_description_europe_2019,
    'europe_2024': political_parties_description_europe_2024,
}
```

```python
resolution = 100
max_score = 10

n_rows = 3
landscape = np.zeros((n_rows, resolution+1))
values = np.arange(0, 10.1, max_score/resolution)
xticks = np.arange(0, resolution+1, resolution/max_score)
xticks_labels = [int(x) for x in xticks/max_score]

for i in range(n_rows):
    landscape[i] = values

for election_id in elections:
    election = elections[election_id]
    my_cmap = cm.get_cmap('RdBu')
    my_norm = colrs.Normalize(vmin = values[0], vmax=values[-1])

    fig = plt.figure(figsize=(10, 60))
    plt.imshow(landscape, cmap=my_cmap, norm=my_norm, interpolation='nearest')

    plt.text(-2, 1, 'Left', fontsize=15, ha='right', va='center', alpha=.85)
    plt.text(0, -2.5, 0, fontsize=12, ha='right', va='center', alpha=.85)

    plt.text(resolution+2, 1, 'Right', fontsize=15, ha='left', va='center', alpha=.85)
    plt.text(resolution, -2.5, 10, fontsize=12, ha='left', va='center', alpha=.85)

    ax = plt.gca()

    for party in election:
        short_name = election[party]['short_name']
        lr_score = election[party]['lr_score']
        votes = election[party]['%votes']
        name = election[party]['name']
        color = election[party]['color']

        x_loc = lr_score*resolution/max_score
        
        radius = votes/100 * 50

        y_loc_abs = 10

        if party in ['RN', 'Ren_coal', 'BE_coal', 'EELV']:

            y_loc = y_loc_abs
            ax.plot([x_loc, x_loc], [n_rows-.5, y_loc], color=color, alpha=.8)
            circle = plt.Circle((x_loc, y_loc+radius), radius, color=color, alpha=.8)
            ax.add_patch(circle)
            ax.text(x_loc, y_loc+2*radius+4, short_name, fontsize=15, ha='center', va='center', color='black', alpha=0.85)

            if party in ['Ren_coal', 'BE_coal']:
                ax.text(x_loc, y_loc+radius, f'{int(votes)}%', fontsize=10+0.7*radius, ha='center', va='center', color='black', alpha=.85)
            else:
                ax.text(x_loc, y_loc+radius, f'{int(votes)}%', fontsize=10+0.7*radius, ha='center', va='center', color='white', alpha=1)
            
            ax.text(x_loc-.2, n_rows+1, f'{lr_score:.1f}', fontsize=12, ha='right', va='center', color='black', alpha=0.85)
        else:
            y_loc = -(y_loc_abs)
            ax.plot([x_loc, x_loc], [-.5, y_loc-n_rows], color=color, alpha=.8)

            circle = plt.Circle((x_loc, y_loc-n_rows-radius), radius, color=color, alpha=.8)
            ax.add_patch(circle)

            if election_id == 'europe_2024' and party == 'FF_coal':                
                ax.text(x_loc+2*radius, y_loc-n_rows-radius, short_name, fontsize=15, ha='left', va='center', color='black', alpha=0.85)
            elif election_id == 'europe_2019' and party == 'EdE_coal':
                ax.text(x_loc-10, y_loc-n_rows-2*radius-4, short_name, fontsize=15, ha='left', va='center', color='black', alpha=0.85)
            else:    
                ax.text(x_loc, y_loc-n_rows-2*radius-4, short_name, fontsize=15, ha='center', va='center', color='black', alpha=0.85)
            
            ax.text(x_loc, y_loc-n_rows-radius, f'{int(votes)}%', fontsize=10+0.7*radius, ha='center', va='center', color='white', alpha=1)
            ax.text(x_loc-.2, -n_rows-.5, f'{lr_score:.1f}', fontsize=12, ha='right', va='center', color='black', alpha=0.85)

    plt.ylim(-40, 45)
    plt.xlim(-20, 120)

    plt.xticks(xticks, xticks_labels)
    plt.axis('off')

    plt.savefig(f'{IMG_DIR}/elections/{election_id}_results_spectrum.pdf', bbox_inches='tight', dpi=200)
    plt.show()
```
